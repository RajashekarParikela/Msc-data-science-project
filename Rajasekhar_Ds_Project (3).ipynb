```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("Columns in df:", list(df.columns))
if 'route_type' not in df.columns:
    if 'origin_destination_country' in df.columns:
        uk_names = {
            'United Kingdom','UK','Great Britain','GB','England','Scotland',
            'Wales','Northern Ireland','United Kingdom of Great Britain and Northern Ireland'
        }
        df['route_type'] = np.where(
            df['origin_destination_country'].astype(str).str.strip().isin(uk_names),
            'Domestic','International'
        )
        print("Derived 'route_type' from 'origin_destination_country'.")
    else:
        print("No 'route_type' or 'origin_destination_country'; proceeding without it.")
candidate_features = [
    'route_type','airline_name','reporting_airport',
    'flights_cancelled_percent','total_late_percent',
    'previous_year_month_average_delay','number_flights_matched'
]
features = [f for f in candidate_features if f in df.columns]
print("Using features:", features)
if 'average_delay_mins' not in df.columns:
    raise ValueError("Target column 'average_delay_mins' not found!")
X = df[features].copy()
y = df['average_delay_mins'].astype(float)
cat_cols = [c for c in X.columns if X[c].dtype == 'object']
if cat_cols:
    for c in cat_cols:
        le = LabelEncoder()
        X[c] = le.fit_transform(X[c].astype(str))
    print("Encoded categorical columns:", cat_cols)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
rf_reg = RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1)
rf_reg.fit(X_train, y_train)
y_pred = rf_reg.predict(X_test)
mae  = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2   = r2_score(y_test, y_pred)
print("\n===== Random Forest Regressor Performance =====")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.3f}")

# Actual vs Predicted
plt.figure(figsize=(7,6))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.6, color='teal')
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         color='red', linestyle='--', label='Perfect Prediction')
plt.title("Actual vs Predicted Delay (minutes)")
plt.xlabel("Actual Delay (minutes)")
plt.ylabel("Predicted Delay (minutes)")
plt.legend()
plt.grid(True)
plt.show()

# Feature Importance
fi = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_reg.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x='Importance', y='Feature', data=fi, palette='viridis')
plt.title("Feature Importance – Random Forest Regressor")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()
print("\nTop features influencing delay:")
display(fi.head(10))

# Line Comparison
plt.figure(figsize=(10,5))
plt.plot(range(50), y_test.values[:50], label='Actual', marker='o')
plt.plot(range(50), y_pred[:50], label='Predicted', marker='x')
plt.title("Actual vs Predicted Delays (Sample of 50)")
plt.xlabel("Sample Index")
plt.ylabel("Delay (minutes)")
plt.legend()
plt.show()


# Correlation Heatmap (Actual vs Predicted)
compare_df = pd.DataFrame({
    'Actual Delay (mins)': y_test,
    'Predicted Delay (mins)': y_pred
})
corr = compare_df.corr()
plt.figure(figsize=(6,5))
sns.heatmap(
    corr, annot=True, cmap='RdYlBu_r', fmt=".2f",
    annot_kws={"size":14,"weight":"bold"},
    linewidths=0.5, cbar_kws={"shrink":0.8,"label":"Correlation Strength"}
)
plt.title("Correlation Heatmap: Actual vs Predicted Delays",
          fontsize=14, fontweight='bold', pad=12)
plt.xticks(fontsize=12, rotation=30, ha='right')
plt.yticks(fontsize=12, rotation=0)
plt.tight_layout()
plt.show()
corr_value = corr.iloc[0,1]
print(f"Correlation between Actual and Predicted Delay: {corr_value:.3f}")
```

```python
#2 – Random Forest Classifier
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)


#Verify target column exists
if 'is_delayed' not in df.columns:
    threshold = 15
    df['is_delayed'] = (df['average_delay_mins'] > threshold).astype(int)
    print(f"Created 'is_delayed' column using {threshold}-minute threshold.")
#Select features
candidate_features = [
    'route_type', 'airline_name', 'reporting_airport',
    'flights_cancelled_percent', 'total_late_percent',
    'previous_year_month_average_delay', 'number_flights_matched'
]
features = [f for f in candidate_features if f in df.columns]
print("Using features:", features)
X = df[features].copy()
y = df['is_delayed']
cat_cols = [c for c in X.columns if X[c].dtype == 'object']
if cat_cols:
    for c in cat_cols:
        le = LabelEncoder()
        X[c] = le.fit_transform(X[c].astype(str))
    print("Encoded categorical columns:", cat_cols)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Train Model
rf_clf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight='balanced'
)
rf_clf.fit(X_train, y_train)

#Predictions & Evaluation
y_pred = rf_clf.predict(X_test)

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("\n===== Random Forest Classifier Performance =====")
print(f"Accuracy : {acc:.3f}")
print(f"Precision: {prec:.3f}")
print(f"Recall   : {rec:.3f}")
print(f"F1-Score : {f1:.3f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['On-time','Delayed']))

#Confusion Matrix
plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test, y_pred),
            annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['On-time','Delayed'],
            yticklabels=['On-time','Delayed'])
plt.title("Confusion Matrix – Flight Delay Classification")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()


#Feature Importance
fi = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_clf.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x='Importance', y='Feature', data=fi, palette='viridis')
plt.title("Feature Importance – Random Forest Classifier")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()

print("\nTop features influencing delay probability:")
display(fi.head(10))
```

```python
# XGBoost Classifier
!pip install xgboost
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score, roc_curve
)
from xgboost import XGBClassifier

#Data Preparation
if 'is_delayed' not in df.columns:
    df['is_delayed'] = (df['average_delay_mins'] > 15).astype(int)
features = [
    'route_type','airline_name','reporting_airport',
    'flights_cancelled_percent','total_late_percent',
    'previous_year_month_average_delay','number_flights_matched'
]
features = [f for f in features if f in df.columns]
print("Using features:", features)
X = df[features].copy()
y = df['is_delayed']
cat_cols = [c for c in X.columns if X[c].dtype == 'object']
for c in cat_cols:
    le = LabelEncoder()
    X[c] = le.fit_transform(X[c].astype(str))
print("Encoded columns:", cat_cols)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Train XGBoost Classifier
xgb_clf = XGBClassifier(
    n_estimators=300,
    learning_rate=0.1,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    use_label_encoder=False,
    eval_metric='logloss'
)
xgb_clf.fit(X_train, y_train)

#Predictions & Evaluation

y_pred = xgb_clf.predict(X_test)
y_proba = xgb_clf.predict_proba(X_test)[:, 1]

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

print("\n===== XGBoost Classifier Performance =====")
print(f"Accuracy : {acc:.3f}")
print(f"Precision: {prec:.3f}")
print(f"Recall   : {rec:.3f}")
print(f"F1-Score : {f1:.3f}")
print(f"AUC      : {auc:.3f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['On-time','Delayed']))

# Confusion Matrix
plt.figure(figsize=(5,4))
sns.heatmap(confusion_matrix(y_test, y_pred),
            annot=True, fmt='d', cmap='Greens', cbar=False,
            xticklabels=['On-time','Delayed'],
            yticklabels=['On-time','Delayed'])
plt.title("Confusion Matrix – XGBoost Classifier")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.tight_layout()
plt.show()


#Feature Importance Plot
xgb_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': xgb_clf.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x='Importance', y='Feature', data=xgb_importance, palette='viridis')
plt.title("Feature Importance – XGBoost Classifier", fontsize=13, fontweight='bold')
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

print("\nTop features influencing delay prediction:")
display(xgb_importance.head(10))
```
